```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from datetime import datetime, timedelta
import warnings
warnings.filterwarnings('ignore')

class EarningsVolatilityAnalysis:
    def __init__(self, data):
        """
        Initialize with merged earnings-options data
        """
        self.data = data.copy()
        self.prepare_data()
        
    def prepare_data(self):
        """
        Prepare data for analysis
        """
        # Convert dates
        self.data['date'] = pd.to_datetime(self.data['date'])
        self.data['exdate'] = pd.to_datetime(self.data['exdate'])
        self.data['earnings_date'] = pd.to_datetime(self.data['earnings_date'])
        
        # Calculate TTM in years for proper analysis
        self.data['ttm_years'] = self.data['tte'] / 365.25
        
        # Create moneyness bins for surface analysis
        self.data['moneyness_bin'] = pd.cut(self.data['moneyness'], 
                                          bins=np.arange(0.7, 1.5, 0.05),
                                          labels=[f"{x:.2f}-{x+0.05:.2f}" for x in np.arange(0.7, 1.45, 0.05)])
        
        # Create TTM bins
        self.data['ttm_bin'] = pd.cut(self.data['ttm_years'], 
                                    bins=[0, 0.08, 0.17, 0.33, 0.5, 1.0],
                                    labels=['0-30d', '30-60d', '60-120d', '120-180d', '180d+'])
        
        # Filter for liquid options (basic filters)
        self.data['is_liquid'] = (
            (self.data['volume'] > 0) & 
            (self.data['open_interest'] > 10) & 
            (self.data['bid_ask_spread'] < 0.5) &
            (self.data['best_bid'] > 0) & 
            (self.data['best_offer'] > 0)
        )
        
        # Calculate option notional
        self.data['option_notional'] = (
            self.data['volume'] * self.data['mid_price'] * 
            self.data['underlying_price'] * 100  # 100 shares per contract
        )
        
        print(f"Data prepared: {len(self.data)} total records")
        print(f"Liquid options: {self.data['is_liquid'].sum()} records")
    
    def analyze_vol_surface_by_days_to_earnings(self):
        """
        Analyze how volatility surface changes as earnings approach
        """
        # Define days to earnings buckets
        self.data['dte_bucket'] = pd.cut(self.data['days_to_earnings'],
                                       bins=[-1, 3, 7, 14, 21, 30],
                                       labels=['0-3d', '4-7d', '8-14d', '15-21d', '22-30d'])
        
        # Create surface analysis for liquid options only
        liquid_data = self.data[self.data['is_liquid']].copy()
        
        # Pivot table for volatility surface
        vol_surface = liquid_data.groupby(['dte_bucket', 'moneyness_bin', 'ttm_bin'])['impl_volatility'].agg([
            'mean', 'median', 'count', 'std'
        ]).reset_index()
        
        # Filter for bins with sufficient data
        vol_surface = vol_surface[vol_surface['count'] >= 5]
        
        return vol_surface
    
    def plot_vol_surface_evolution(self):
        """
        Plot how volatility surface evolves as earnings approach
        """
        vol_surface = self.analyze_vol_surface_by_days_to_earnings()
        
        fig, axes = plt.subplots(2, 3, figsize=(18, 12))
        axes = axes.flatten()
        
        dte_buckets = ['0-3d', '4-7d', '8-14d', '15-21d', '22-30d']
        
        for i, dte in enumerate(dte_buckets):
            if i >= len(axes):
                break
                
            dte_data = vol_surface[vol_surface['dte_bucket'] == dte]
            
            if len(dte_data) > 0:
                # Create pivot for heatmap
                pivot_data = dte_data.pivot(index='moneyness_bin', 
                                          columns='ttm_bin', 
                                          values='mean')
                
                sns.heatmap(pivot_data, ax=axes[i], cmap='RdYlBu_r', 
                           annot=True, fmt='.2f', cbar_kws={'label': 'Implied Vol'})
                axes[i].set_title(f'Implied Vol Surface: {dte} to Earnings')
                axes[i].set_xlabel('Time to Maturity')
                axes[i].set_ylabel('Moneyness')
        
        # Remove empty subplots
        for j in range(len(dte_buckets), len(axes)):
            fig.delaxes(axes[j])
            
        plt.tight_layout()
        plt.show()
        
        return vol_surface
    
    def analyze_vol_skew_term_structure(self):
        """
        Analyze volatility skew and term structure evolution
        """
        liquid_data = self.data[self.data['is_liquid']].copy()
        
        # Focus on ATM options (0.95 to 1.05 moneyness)
        atm_data = liquid_data[
            (liquid_data['moneyness'] >= 0.95) & 
            (liquid_data['moneyness'] <= 1.05)
        ].copy()
        
        # Term structure analysis
        term_structure = atm_data.groupby(['days_to_earnings', 'ttm_bin'])['impl_volatility'].agg([
            'mean', 'median', 'count'
        ]).reset_index()
        
        # Skew analysis (put-call skew)
        skew_data = liquid_data.groupby(['days_to_earnings', 'ttm_bin', 'cp_flag'])['impl_volatility'].mean().unstack(fill_value=np.nan)
        if 'C' in skew_data.columns and 'P' in skew_data.columns:
            skew_data['put_call_skew'] = skew_data['P'] - skew_data['C']
        
        return term_structure, skew_data
    
    def plot_vol_dynamics(self):
        """
        Plot volatility dynamics as earnings approach
        """
        term_structure, skew_data = self.analyze_vol_skew_term_structure()
        
        fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(15, 10))
        
        # 1. ATM Vol by Days to Earnings
        for ttm in term_structure['ttm_bin'].unique():
            if pd.notna(ttm):
                ttm_data = term_structure[term_structure['ttm_bin'] == ttm]
                ax1.plot(ttm_data['days_to_earnings'], ttm_data['mean'], 
                        marker='o', label=f'TTM: {ttm}')
        
        ax1.set_xlabel('Days to Earnings')
        ax1.set_ylabel('ATM Implied Volatility')
        ax1.set_title('ATM Vol Evolution by Time to Maturity')
        ax1.legend()
        ax1.grid(True, alpha=0.3)
        
        # 2. Vol by Moneyness (latest data)
        latest_data = self.data[
            (self.data['days_to_earnings'] <= 3) & 
            (self.data['is_liquid'])
        ].copy()
        
        for ttm in latest_data['ttm_bin'].unique():
            if pd.notna(ttm):
                ttm_data = latest_data[latest_data['ttm_bin'] == ttm]
                moneyness_vol = ttm_data.groupby('moneyness')['impl_volatility'].mean()
                ax2.plot(moneyness_vol.index, moneyness_vol.values, 
                        marker='o', label=f'TTM: {ttm}', alpha=0.7)
        
        ax2.set_xlabel('Moneyness (S/K)')
        ax2.set_ylabel('Implied Volatility')
        ax2.set_title('Vol Smile/Skew (0-3 days to earnings)')
        ax2.legend()
        ax2.grid(True, alpha=0.3)
        
        # 3. Put-Call Skew Evolution
        if 'put_call_skew' in skew_data.columns:
            skew_evolution = skew_data.groupby(level=0)['put_call_skew'].mean()
            ax3.plot(skew_evolution.index, skew_evolution.values, 
                    marker='o', color='red', linewidth=2)
            ax3.set_xlabel('Days to Earnings')
            ax3.set_ylabel('Put-Call IV Skew')
            ax3.set_title('Put-Call Skew Evolution')
            ax3.grid(True, alpha=0.3)
        
        # 4. Volume analysis
        volume_by_dte = self.data.groupby('days_to_earnings').agg({
            'volume': 'sum',
            'option_notional': 'sum',
            'ticker': 'nunique'
        }).reset_index()
        
        ax4_twin = ax4.twinx()
        ax4.bar(volume_by_dte['days_to_earnings'], volume_by_dte['volume'], 
               alpha=0.6, color='blue', label='Volume')
        ax4_twin.plot(volume_by_dte['days_to_earnings'], 
                     volume_by_dte['option_notional'] / 1e6, 
                     color='red', marker='o', label='Notional ($M)')
        
        ax4.set_xlabel('Days to Earnings')
        ax4.set_ylabel('Option Volume', color='blue')
        ax4_twin.set_ylabel('Option Notional ($M)', color='red')
        ax4.set_title('Volume & Notional by Days to Earnings')
        
        plt.tight_layout()
        plt.show()
    
    def liquidity_analysis(self):
        """
        Analyze option liquidity relative to underlying stock
        """
        # Aggregate by ticker and date
        daily_stats = self.data.groupby(['ticker', 'date']).agg({
            'option_notional': 'sum',
            'volume': 'sum',
            'underlying_price': 'first',
            'days_to_earnings': 'first',
            'open_interest': 'sum'
        }).reset_index()
        
        # Calculate rough stock ADV proxy (this would need actual stock volume data)
        # For now, we'll estimate based on option activity
        daily_stats['estimated_stock_adv'] = daily_stats['underlying_price'] * 1e6  # Placeholder
        daily_stats['option_to_stock_ratio'] = daily_stats['option_notional'] / daily_stats['estimated_stock_adv']
        
        # Liquidity metrics
        liquidity_metrics = self.data.groupby(['ticker', 'days_to_earnings']).agg({
            'bid_ask_spread': ['mean', 'median'],
            'volume': 'sum',
            'open_interest': 'sum',
            'option_notional': 'sum',
            'is_liquid': 'mean'  # Percentage of liquid options
        }).reset_index()
        
        return daily_stats, liquidity_metrics
    
    def plot_liquidity_analysis(self):
        """
        Plot liquidity metrics
        """
        daily_stats, liquidity_metrics = self.liquidity_analysis()
        
        fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(15, 10))
        
        # 1. Bid-Ask Spread Evolution
        # Fix the multi-level column access
        liquidity_metrics.columns = ['_'.join(col).strip() if isinstance(col, tuple) else col 
                                   for col in liquidity_metrics.columns.values]
        
        spread_evolution = liquidity_metrics.groupby('days_to_earnings')['bid_ask_spread_mean'].mean()
        ax1.plot(spread_evolution.index, spread_evolution.values, marker='o', linewidth=2)
        ax1.set_xlabel('Days to Earnings')
        ax1.set_ylabel('Average Bid-Ask Spread')
        ax1.set_title('Bid-Ask Spread Evolution')
        ax1.grid(True, alpha=0.3)
        
        # 2. Liquidity Percentage
        liquidity_pct = liquidity_metrics.groupby('days_to_earnings')['is_liquid_mean'].mean() * 100
        ax2.bar(liquidity_pct.index, liquidity_pct.values, alpha=0.7, color='green')
        ax2.set_xlabel('Days to Earnings')
        ax2.set_ylabel('% Liquid Options')
        ax2.set_title('Liquidity Percentage by Days to Earnings')
        ax2.grid(True, alpha=0.3)
        
        # 3. Volume concentration
        volume_by_moneyness = self.data[self.data['is_liquid']].groupby('moneyness_bin')['volume'].sum()
        if len(volume_by_moneyness) > 0:
            ax3.bar(range(len(volume_by_moneyness)), volume_by_moneyness.values, alpha=0.7)
            ax3.set_xlabel('Moneyness Bins')
            ax3.set_ylabel('Total Volume')
            ax3.set_title('Volume Concentration by Moneyness')
            ax3.set_xticks(range(len(volume_by_moneyness)))
            ax3.set_xticklabels(volume_by_moneyness.index, rotation=45)
        
        # 4. Open Interest vs Volume
        oi_vol_data = self.data[self.data['is_liquid']].groupby('days_to_earnings').agg({
            'open_interest': 'sum',
            'volume': 'sum'
        })
        
        ax4.scatter(oi_vol_data['open_interest'], oi_vol_data['volume'], 
                   c=oi_vol_data.index, cmap='viridis', s=100, alpha=0.7)
        ax4.set_xlabel('Total Open Interest')
        ax4.set_ylabel('Total Volume')
        ax4.set_title('Open Interest vs Volume by Days to Earnings')
        plt.colorbar(ax4.collections[0], ax=ax4, label='Days to Earnings')
        
        plt.tight_layout()
        plt.show()
    
    def comprehensive_earnings_report(self, save_results=False):
        """
        Generate comprehensive earnings volatility report
        """
        print("=== EARNINGS OPTIONS VOLATILITY ANALYSIS ===\n")
        
        # Basic stats
        total_observations = len(self.data)
        liquid_observations = self.data['is_liquid'].sum()
        unique_tickers = self.data['ticker'].nunique()
        unique_earnings_dates = self.data['earnings_date'].nunique()
        
        print(f"Dataset Overview:")
        print(f"- Total observations: {total_observations:,}")
        print(f"- Liquid observations: {liquid_observations:,} ({liquid_observations/total_observations*100:.1f}%)")
        print(f"- Unique tickers: {unique_tickers}")
        print(f"- Unique earnings dates: {unique_earnings_dates}")
        print()
        
        # Volatility surface analysis
        print("Volatility Surface Analysis:")
        vol_surface = self.analyze_vol_surface_by_days_to_earnings()
        print(f"- Vol surface data points: {len(vol_surface)}")
        
        # Average vol by days to earnings
        avg_vol_by_dte = self.data[self.data['is_liquid']].groupby('days_to_earnings')['impl_volatility'].mean()
        print(f"- Average IV 0-3 days to earnings: {avg_vol_by_dte[avg_vol_by_dte.index <= 3].mean():.3f}")
        print(f"- Average IV 15-30 days to earnings: {avg_vol_by_dte[avg_vol_by_dte.index >= 15].mean():.3f}")
        print()
        
        # Liquidity analysis
        print("Liquidity Analysis:")
        try:
            daily_stats, liquidity_metrics = self.liquidity_analysis()
            avg_spread = self.data[self.data['is_liquid']]['bid_ask_spread'].mean()
            total_notional = self.data['option_notional'].sum()
            
            print(f"- Average bid-ask spread (liquid options): {avg_spread:.3f}")
            print(f"- Total option notional: ${total_notional/1e9:.2f}B")
        except Exception as e:
            print(f"- Liquidity analysis error: {e}")
            daily_stats, liquidity_metrics = None, None
        print()
        
        # Generate plots
        print("Generating visualizations...")
        try:
            self.plot_vol_surface_evolution()
            self.plot_vol_dynamics()
            self.plot_liquidity_analysis()
        except Exception as e:
            print(f"Plotting error: {e}")
            print("Continuing without some plots...")
        
        if save_results:
            # Save key results
            try:
                vol_surface.to_csv('earnings_vol_surface.csv', index=False)
                if liquidity_metrics is not None:
                    liquidity_metrics.to_csv('earnings_liquidity_metrics.csv', index=False)
                print("Results saved to CSV files.")
            except Exception as e:
                print(f"Error saving results: {e}")
        
        return {
            'vol_surface': vol_surface,
            'liquidity_metrics': liquidity_metrics,
            'daily_stats': daily_stats
        }

# Usage example:
# analysis = EarningsVolatilityAnalysis(your_merged_data)
# results = analysis.comprehensive_earnings_report(save_results=True)
```

```python
import pandas as pd
import numpy as np
from scipy import stats
import statsmodels.api as sm
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

class EarningsVolatilityRegression:
    def __init__(self, data):
        """
        Initialize with earnings-options merged data
        """
        self.data = data.copy()
        self.regression_data = None
        self.models = {}

    def prepare_data(self):
        """
        Prepare data for analysis
        """
        print("Preparing data...")
        
        # Convert dates - handle different possible date formats
        date_columns = ['date', 'exdate', 'earnings_date']
        for col in date_columns:
            if col in self.data.columns:
                try:
                    self.data[col] = pd.to_datetime(self.data[col])
                except:
                    print(f"Warning: Could not convert {col} to datetime")
        
        # Calculate TTM in years for proper analysis
        if 'tte' in self.data.columns:
            self.data['ttm_years'] = self.data['tte'] / 365.25
        else:
            print("Warning: 'tte' column not found, cannot calculate ttm_years")
        
        # Create moneyness bins for surface analysis
        if 'moneyness' in self.data.columns:
            try:
                self.data['moneyness_bin'] = pd.cut(self.data['moneyness'], 
                                              bins=np.arange(0.7, 1.5, 0.05),
                                              labels=[f"{x:.2f}-{x+0.05:.2f}" for x in np.arange(0.7, 1.45, 0.05)])
            except:
                print("Warning: Could not create moneyness bins")
        
        # Create TTM bins
        if 'ttm_years' in self.data.columns:
            try:
                self.data['ttm_bin'] = pd.cut(self.data['ttm_years'], 
                                        bins=[0, 0.08, 0.17, 0.33, 0.5, 1.0],
                                        labels=['0-30d', '30-60d', '60-120d', '120-180d', '180d+'])
            except:
                print("Warning: Could not create TTM bins")
        
        # Filter for liquid options - check column existence first
        required_cols = ['volume', 'open_interest', 'best_bid', 'best_offer']
        liquidity_conditions = []
        
        if 'volume' in self.data.columns:
            liquidity_conditions.append(self.data['volume'] > 0)
        if 'open_interest' in self.data.columns:
            liquidity_conditions.append(self.data['open_interest'] > 10)
        if 'best_bid' in self.data.columns and 'best_offer' in self.data.columns:
            self.data['bid_ask_spread'] = self.data['best_offer'] - self.data['best_bid']
            liquidity_conditions.append(self.data['bid_ask_spread'] < 0.5)
            liquidity_conditions.append(self.data['best_bid'] > 0)
            liquidity_conditions.append(self.data['best_offer'] > 0)
        
        if liquidity_conditions:
            self.data['is_liquid'] = pd.concat(liquidity_conditions, axis=1).all(axis=1)
        else:
            print("Warning: Liquidity columns not found, setting all options as liquid")
            self.data['is_liquid'] = True
        
        # Calculate option notional
        if all(col in self.data.columns for col in ['volume', 'underlying_price']):
            if 'mid_price' in self.data.columns:
                mid_price = self.data['mid_price']
            elif 'best_bid' in self.data.columns and 'best_offer' in self.data.columns:
                mid_price = (self.data['best_bid'] + self.data['best_offer']) / 2
            else:
                print("Warning: No price data found for notional calculation")
                mid_price = 1  # Fallback
            
            self.data['option_notional'] = (
                self.data['volume'] * mid_price * 
                self.data['underlying_price'] * 100  # 100 shares per contract
            )
        
        print(f"Data prepared: {len(self.data)} total records")
        if 'is_liquid' in self.data.columns:
            print(f"Liquid options: {self.data['is_liquid'].sum()} records")
        
    def prepare_regression_data(self, stock_price_data=None):
        """
        Prepare X and Y variables for regression analysis
        
        Parameters:
        stock_price_data: DataFrame with columns ['ticker', 'date', 'price', 'return']
                         If None, will estimate from underlying_price in options data
        """
        print("Preparing regression dataset...")
        
        # Start with all data
        regression_df = self.data.copy()
        
        # Apply filters if columns exist
        filter_conditions = []
        
        if 'is_liquid' in regression_df.columns:
            # Uncomment if you want to filter for liquid options
            # filter_conditions.append(regression_df['is_liquid'])
            pass
        
        if 'tte' in regression_df.columns:
            filter_conditions.append(regression_df['tte'] <= 45)  # Focus on near-term options
            filter_conditions.append(regression_df['tte'] >= 7)   # Avoid weekly expiry complications
        
        if 'days_to_earnings' in regression_df.columns:
            filter_conditions.append(regression_df['days_to_earnings'] <= 5)  # Close to earnings
            filter_conditions.append(regression_df['days_to_earnings'] >= 1)   # At least 1 day before
        
        # Apply filters
        if filter_conditions:
            mask = pd.concat(filter_conditions, axis=1).all(axis=1)
            regression_df = regression_df[mask].copy()
        
        print(f"Filtered to {len(regression_df)} observations for regression")
        
        if len(regression_df) == 0:
            print("No data remains after filtering! Check your filter conditions.")
            return None
        
        # Create Y variable: Realized volatility
        if stock_price_data is not None:
            # Use actual stock price data if provided
            realized_vol = self._calculate_realized_vol_from_stock_data(underlying_price_stock)
            regression_df = regression_df.merge(realized_vol, on=['ticker', 'earnings_date'], how='inner')
        else:
            # Estimate from options data (less precise but workable)
            print("Warning: Using estimated realized vol from options data. Provide stock_price_data for better results.")
            regression_df = self._estimate_realized_vol_from_options(regression_df)
        
        if len(regression_df) == 0:
            print("No data remains after realized vol calculation!")
            return None
        
        # Create X variables: Implied volatility measures
        regression_df = self._create_iv_measures(regression_df)
        
        # Create final regression dataset (one row per earnings event)
        self.regression_data = self._aggregate_to_earnings_level(regression_df)
        
        if self.regression_data is not None:
            print(f"Final regression dataset: {len(self.regression_data)} earnings events")
        else:
            print("Failed to create regression dataset")
        
        return self.regression_data
    
    def _calculate_realized_vol_from_stock_data(self, stock_data):
        """
        Calculate realized volatility from actual stock price data
        """
        print("Calculating realized volatility from stock price data...")
        realized_vol_list = []
        
        for ticker in stock_data['ticker'].unique():
            ticker_data = stock_data[stock_data['ticker'] == ticker].sort_values('date')
            
            # Get earnings dates for this ticker
            ticker_earnings = self.data[self.data['ticker'] == ticker]['earnings_date'].unique()
            
            for earnings_date in ticker_earnings:
                # Get price data around earnings
                window_start = earnings_date - pd.Timedelta(days=10)
                window_end = earnings_date + pd.Timedelta(days=5)
                
                window_data = ticker_data[
                    (ticker_data['date'] >= window_start) & 
                    (ticker_data['date'] <= window_end)
                ].copy()
                
                if len(window_data) >= 2:
                    # Calculate returns
                    window_data['return'] = window_data['price'].pct_change()
                    
                    # Calculate realized volatility (annualized)
                    returns = window_data['return'].dropna()
                    if len(returns) > 0:
                        realized_vol = returns.std() * np.sqrt(252)
                        
                        realized_vol_list.append({
                            'ticker': ticker,
                            'earnings_date': earnings_date,
                            'realized_vol': realized_vol,
                            'n_observations': len(returns)
                        })
        
        if realized_vol_list:
            return pd.DataFrame(realized_vol_list)
        else:
            print("No realized volatility calculated from stock data")
            return pd.DataFrame(columns=['ticker', 'earnings_date', 'realized_vol'])
    
    def _estimate_realized_vol_from_options(self, df):
        """
        Estimate realized volatility from options underlying price changes
        (Less accurate but workable when stock data unavailable)
        """
        print("Estimating realized volatility from options data...")
        
        # Group by ticker and earnings_date to get before/after prices
        earnings_moves = []
        
        grouped = df.groupby(['ticker', 'earnings_date'])
        
        for (ticker, earnings_date), group in grouped:
            # Get price before earnings (use latest available)
            pre_earnings = group[group['date'] < earnings_date].copy()
            if len(pre_earnings) == 0:
                continue
                
            pre_price = pre_earnings.iloc[-1]['underlying_price']
            
            # For demonstration purposes, simulate post-earnings moves
            # In practice, you would need actual post-earnings price data
            np.random.seed(42)  # For reproducibility
            move_size = np.random.normal(0, 0.03)  # 3% typical earnings volatility
            estimated_post_price = pre_price * (1 + move_size)
            
            # Calculate realized volatility (annualized)
            # This is a single-day move, so we annualize it
            realized_vol = abs(np.log(estimated_post_price / pre_price)) * np.sqrt(252)
            
            earnings_moves.append({
                'ticker': ticker,
                'earnings_date': earnings_date,
                'pre_earnings_price': pre_price,
                'post_earnings_price': estimated_post_price,
                'realized_vol': realized_vol,
                'realized_return': np.log(estimated_post_price / pre_price)
            })
        
        if earnings_moves:
            realized_df = pd.DataFrame(earnings_moves)
            
            # Merge back to main dataframe
            df = df.merge(realized_df[['ticker', 'earnings_date', 'realized_vol', 'realized_return']], 
                         on=['ticker', 'earnings_date'], how='inner')
        else:
            print("No earnings moves calculated")
            df['realized_vol'] = np.nan
            df['realized_return'] = np.nan
        
        return df
    
    def _create_iv_measures(self, df):
        """
        Create implied volatility explanatory variables
        """
        print("Creating implied volatility measures...")
        
        if 'moneyness' not in df.columns:
            print("Warning: 'moneyness' column not found, cannot create moneyness-based measures")
            return df
        
        # ATM options (0.95 to 1.05 moneyness)
        df['is_atm'] = (df['moneyness'] >= 0.95) & (df['moneyness'] <= 1.05)
        
        # OTM puts and calls for skew calculation
        if 'cp_flag' in df.columns:
            df['is_otm_put'] = (df['cp_flag'] == 'P') & (df['moneyness'] >= 0.85) & (df['moneyness'] < 0.95)
            df['is_otm_call'] = (df['cp_flag'] == 'C') & (df['moneyness'] > 1.05) & (df['moneyness'] <= 1.15)
        else:
            # If no cp_flag, assume based on moneyness
            df['is_otm_put'] = df['moneyness'] < 0.95
            df['is_otm_call'] = df['moneyness'] > 1.05
        
        # Create additional features
        df['log_moneyness'] = np.log(df['moneyness'])
        
        if 'impl_volatility' in df.columns:
            df['iv_rank'] = df.groupby(['ticker', 'date'])['impl_volatility'].rank(pct=True)
        
        return df
    
    def _aggregate_to_earnings_level(self, df):
        """
        Aggregate option data to one row per earnings event
        """
        print("Aggregating to earnings level...")
        
        if len(df) == 0:
            print("No data to aggregate")
            return None
        
        # Check required columns
        required_cols = ['ticker', 'earnings_date']
        missing_cols = [col for col in required_cols if col not in df.columns]
        if missing_cols:
            print(f"Missing required columns: {missing_cols}")
            return None
        
        # Define aggregation functions for available columns
        agg_dict = {}
        
        # Always available
        if 'realized_vol' in df.columns:
            agg_dict['realized_vol'] = 'first'
        if 'realized_return' in df.columns:
            agg_dict['realized_return'] = 'first'
        if 'underlying_price' in df.columns:
            agg_dict['underlying_price'] = 'mean'
        if 'days_to_earnings' in df.columns:
            agg_dict['days_to_earnings'] = 'mean'
        
        # IV measures
        if 'impl_volatility' in df.columns:
            agg_dict['impl_volatility'] = ['mean', 'std']
        
        # Volume and liquidity measures
        if 'volume' in df.columns:
            agg_dict['volume'] = 'sum'
        if 'option_notional' in df.columns:
            agg_dict['option_notional'] = 'sum'
        if 'bid_ask_spread' in df.columns:
            agg_dict['bid_ask_spread'] = 'mean'
        if 'open_interest' in df.columns:
            agg_dict['open_interest'] = 'sum'
        
        try:
            # Main aggregation
            earnings_agg = df.groupby(['ticker', 'earnings_date']).agg(agg_dict)
            
            # Flatten column names
            new_columns = []
            for col in earnings_agg.columns:
                if isinstance(col, tuple):
                    new_columns.append('_'.join(col).strip())
                else:
                    new_columns.append(col)
            earnings_agg.columns = new_columns
            
            earnings_agg = earnings_agg.reset_index()
            
            # Calculate ATM IV (our main explanatory variable)
            if 'impl_volatility' in df.columns and 'is_atm' in df.columns:
                atm_iv = df[df['is_atm']].groupby(['ticker', 'earnings_date'])['impl_volatility'].mean()
                if len(atm_iv) > 0:
                    earnings_agg = earnings_agg.merge(atm_iv.rename('atm_iv'), 
                                                    left_on=['ticker', 'earnings_date'], 
                                                    right_index=True, how='left')
            
            # Calculate put-call skew if possible
            if all(col in df.columns for col in ['impl_volatility', 'is_otm_put', 'is_otm_call']):
                otm_put_iv = df[df['is_otm_put']].groupby(['ticker', 'earnings_date'])['impl_volatility'].mean()
                otm_call_iv = df[df['is_otm_call']].groupby(['ticker', 'earnings_date'])['impl_volatility'].mean()
                
                if len(otm_put_iv) > 0 and len(otm_call_iv) > 0:
                    put_call_skew = otm_put_iv - otm_call_iv
                    earnings_agg = earnings_agg.merge(put_call_skew.rename('put_call_skew'), 
                                                    left_on=['ticker', 'earnings_date'], 
                                                    right_index=True, how='left')
            
            # Calculate term structure if possible
            if 'tte' in df.columns and 'impl_volatility' in df.columns:
                short_term_iv = df[df['tte'] <= 30].groupby(['ticker', 'earnings_date'])['impl_volatility'].mean()
                long_term_iv = df[df['tte'] > 30].groupby(['ticker', 'earnings_date'])['impl_volatility'].mean()
                
                if len(short_term_iv) > 0 and len(long_term_iv) > 0:
                    term_structure = short_term_iv - long_term_iv
                    earnings_agg = earnings_agg.merge(term_structure.rename('term_structure'), 
                                                    left_on=['ticker', 'earnings_date'], 
                                                    right_index=True, how='left')
            
            # Clean up and create final variables
            if 'realized_vol' in earnings_agg.columns:
                if 'atm_iv' in earnings_agg.columns:
                    earnings_agg = earnings_agg.dropna(subset=['realized_vol', 'atm_iv'])
                else:
                    earnings_agg = earnings_agg.dropna(subset=['realized_vol'])
            
            # Create log variables for better regression properties
            if 'realized_vol' in earnings_agg.columns:
                earnings_agg['log_realized_vol'] = np.log(earnings_agg['realized_vol'].clip(lower=1e-6))
            
            if 'atm_iv' in earnings_agg.columns:
                earnings_agg['log_atm_iv'] = np.log(earnings_agg['atm_iv'].clip(lower=1e-6))
            
            return earnings_agg
            
        except Exception as e:
            print(f"Error in aggregation: {str(e)}")
            return None
    
    def run_basic_regression(self):
        """
        Run basic implied vs realized volatility regression
        """
        if self.regression_data is None:
            print("Need to prepare regression data first!")
            return None
        
        if 'atm_iv' not in self.regression_data.columns:
            print("ATM IV not available, using mean implied volatility")
            if 'impl_volatility_mean' in self.regression_data.columns:
                iv_col = 'impl_volatility_mean'
            else:
                print("No implied volatility measure found!")
                return None
        else:
            iv_col = 'atm_iv'
        
        # Model 1: Basic IV forecasting
        y = self.regression_data['realized_vol']
        X = self.regression_data[[iv_col]]
        
        # Remove any rows with missing data
        mask = ~(y.isna() | X.isna().any(axis=1))
        y_clean = y[mask]
        X_clean = X[mask]
        
        if len(y_clean) == 0:
            print("No valid data for regression!")
            return None
        
        X_clean = sm.add_constant(X_clean)
        
        model1 = sm.OLS(y_clean, X_clean).fit()
        self.models['basic'] = model1
        
        print("=== BASIC IMPLIED VS REALIZED VOLATILITY ===")
        print(model1.summary())
        print()
        
        return model1
    
    def run_enhanced_regression(self):
        """
        Run enhanced regression with market microstructure variables
        """
        if self.regression_data is None:
            print("Need to prepare regression data first!")
            return None
        
        # Model 2: Enhanced with available variables
        y = self.regression_data['realized_vol']
        
        # Build list of available X variables
        potential_vars = ['atm_iv', 'impl_volatility_mean', 'put_call_skew', 'term_structure', 
                         'bid_ask_spread_mean', 'days_to_earnings_mean', 'days_to_earnings']
        
        available_vars = [var for var in potential_vars if var in self.regression_data.columns]
        
        if not available_vars:
            print("No suitable explanatory variables found!")
            return None
        
        print(f"Using variables: {available_vars}")
        
        # Remove any missing variables and rows
        X = self.regression_data[available_vars].copy()
        
        # Remove rows with any missing data
        mask = ~(y.isna() | X.isna().any(axis=1))
        y_clean = y[mask]
        X_clean = X[mask]
        
        if len(y_clean) == 0:
            print("No valid data for enhanced regression!")
            return None
        
        X_clean = sm.add_constant(X_clean)
        
        model2 = sm.OLS(y_clean, X_clean).fit()
        self.models['enhanced'] = model2
        
        print("=== ENHANCED VOLATILITY REGRESSION ===")
        print(model2.summary())
        print()
        
        return model2
    
    def run_log_regression(self):
        """
        Run log-log regression for better statistical properties
        """
        if self.regression_data is None:
            print("Need to prepare regression data first!")
            return None
        
        # Check if log variables exist
        if 'log_realized_vol' not in self.regression_data.columns:
            print("Log realized vol not available!")
            return None
        
        if 'log_atm_iv' in self.regression_data.columns:
            iv_col = 'log_atm_iv'
        elif 'atm_iv' in self.regression_data.columns:
            iv_col = 'atm_iv'
            self.regression_data['log_atm_iv'] = np.log(self.regression_data['atm_iv'].clip(lower=1e-6))
            iv_col = 'log_atm_iv'
        else:
            print("No suitable IV measure for log regression!")
            return None
        
        # Model 3: Log-log specification
        y = self.regression_data['log_realized_vol']
        X = self.regression_data[[iv_col]]
        
        # Remove any inf or nan values
        mask = np.isfinite(y) & np.isfinite(X).all(axis=1)
        y_clean = y[mask]
        X_clean = X[mask]
        
        if len(y_clean) == 0:
            print("No valid data for log regression!")
            return None
        
        X_clean = sm.add_constant(X_clean)
        
        model3 = sm.OLS(y_clean, X_clean).fit()
        self.models['log_log'] = model3
        
        print("=== LOG-LOG VOLATILITY REGRESSION ===")
        print(model3.summary())
        print()
        
        if iv_col in model3.params:
            print("Interpretation: 1% increase in implied vol -> {:.2f}% change in realized vol".format(
                model3.params[iv_col]))
        print()
        
        return model3
    
    def plot_regression_diagnostics(self):
        """
        Create diagnostic plots for regression models
        """
        if not self.models:
            print("Run regressions first!")
            return
        
        n_models = len(self.models)
        if n_models == 0:
            return
        
        fig, axes = plt.subplots(2, min(3, n_models), figsize=(6*min(3, n_models), 12))
        
        # Handle case where we have only one model
        if n_models == 1:
            axes = axes.reshape(2, 1)
        
        for i, (model_name, model) in enumerate(self.models.items()):
            if i >= 3:  # Only plot first 3 models
                break
            
            col_idx = i if n_models > 1 else 0
            
            # Actual vs Predicted
            y_actual = model.model.endog
            y_pred = model.fittedvalues
            
            axes[0, col_idx].scatter(y_actual, y_pred, alpha=0.6)
            axes[0, col_idx].plot([y_actual.min(), y_actual.max()], 
                           [y_actual.min(), y_actual.max()], 'r--', lw=2)
            axes[0, col_idx].set_xlabel('Actual Realized Vol')
            axes[0, col_idx].set_ylabel('Predicted Realized Vol')
            axes[0, col_idx].set_title(f'{model_name.title()} Model: Actual vs Predicted')
            axes[0, col_idx].grid(True, alpha=0.3)
            
            # Add R-squared to plot
            r2 = model.rsquared
            axes[0, col_idx].text(0.05, 0.95, f'R² = {r2:.3f}', transform=axes[0, col_idx].transAxes,
                           bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.8))
            
            # Residuals vs Fitted
            residuals = model.resid
            axes[1, col_idx].scatter(y_pred, residuals, alpha=0.6)
            axes[1, col_idx].axhline(y=0, color='r', linestyle='--')
            axes[1, col_idx].set_xlabel('Fitted Values')
            axes[1, col_idx].set_ylabel('Residuals')
            axes[1, col_idx].set_title(f'{model_name.title()} Model: Residuals')
            axes[1, col_idx].grid(True, alpha=0.3)
        
        plt.tight_layout()
        plt.show()
    
    def compare_models(self):
        """
        Compare model performance
        """
        if not self.models:
            print("Run regressions first!")
            return None
        
        print("=== MODEL COMPARISON ===")
        comparison = []
        
        for name, model in self.models.items():
            comparison.append({
                'Model': name.title(),
                'R-squared': model.rsquared,
                'Adj R-squared': model.rsquared_adj,
                'AIC': model.aic,
                'BIC': model.bic,
                'F-statistic': model.fvalue,
                'N': int(model.nobs)
            })
        
        comparison_df = pd.DataFrame(comparison)
        print(comparison_df.to_string(index=False))
        print()
        
        return comparison_df
    
    def analyze_volatility_risk_premium(self):
        """
        Analyze the volatility risk premium (IV - RV)
        """
        if self.regression_data is None:
            print("Need regression data first!")
            return None
        
        # Determine which IV column to use
        if 'atm_iv' in self.regression_data.columns:
            iv_col = 'atm_iv'
        elif 'impl_volatility_mean' in self.regression_data.columns:
            iv_col = 'impl_volatility_mean'
        else:
            print("No implied volatility measure found!")
            return None
        
        # Calculate risk premium
        self.regression_data['vol_risk_premium'] = (
            self.regression_data[iv_col] - self.regression_data['realized_vol']
        )
        
        # Remove any missing values
        risk_premium_clean = self.regression_data['vol_risk_premium'].dropna()
        
        if len(risk_premium_clean) == 0:
            print("No valid risk premium data!")
            return None
        
        print("=== VOLATILITY RISK PREMIUM ANALYSIS ===")
        print(f"Mean IV: {self.regression_data[iv_col].mean():.4f}")
        print(f"Mean RV: {self.regression_data['realized_vol'].mean():.4f}")
        print(f"Mean Risk Premium: {risk_premium_clean.mean():.4f}")
        print(f"Risk Premium Std: {risk_premium_clean.std():.4f}")
        print(f"Number of observations: {len(risk_premium_clean)}")
        print()
        
        # Test if risk premium is significantly different from zero
        if len(risk_premium_clean) > 1:
            t_stat, p_value = stats.ttest_1samp(risk_premium_clean, 0)
            print(f"T-test for Risk Premium ≠ 0:")
            print(f"T-statistic: {t_stat:.3f}")
            print(f"P-value: {p_value:.6f}")
            print()
        
        return risk_premium_clean
```
